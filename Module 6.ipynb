### Assign

What is the profit margin for each movie?

```python
movies['profit'] = movies['US_Gross'] - movies['Production_Budget']
movies
```

```python
movies2 = movies.assign(profit = movies['US_Gross'] - movies['Production_Budget'])
movies2.head()
```

What is the profit ratio for each movie?

```python
movies3 = movies2.assign(profit_ratio = movies2['US_Gross'] / movies2['Production_Budget'])
movies3.head()
```

```python
profit_ratio_chart = alt.Chart(movies3).mark_circle().encode(
        x=alt.X("Worldwide_Gross"),
    y=alt.Y("profit_ratio"),
    color="Major_Genre"
)

profit_ratio_chart
```

```python
### What's the outlier?

movies3[movies3['profit_ratio'] > 6000]
```

```python
### How does is look with a log scale?

profit_ratio_chart_log = alt.Chart(movies3).mark_circle().encode(
    x=alt.X("Worldwide_Gross"),
    y=alt.Y("profit_ratio", axis=alt.Axis(tickCount=40),
           # scale=alt.Scale(type="log", base=10)
           scale=alt.Scale(type="symlog")
           ),
    tooltip=[
        alt.Tooltip("Title:N"),
        alt.Tooltip('Worldwide_Gross', format='$s'),
        alt.Tooltip("profit_ratio", format='.2f')
                    ]
    # color="Major_Genre"
)

profit_ratio_chart_log
```

### Merging dataframes

How can we get the counts of each genre in the dataframe?

Step 1: Get the counts of each genre:

```python
genre_counts = movies3.value_counts("Major_Genre")
genre_counts
```

Turn it into a dataframe

```python
genre_counts_df = genre_counts.reset_index()
genre_counts_df
```

```python
genre_counts_df.columns =['a', 'b']
genre_counts_df
```

Rename the columns

```python
genre_counts_df.columns = ['Major_Genre', 'Genre_counts']
# genre_counts_df = genre_counts_df[genre_counts_df['Genre']=="Drama"]
genre_counts_df
```

Merge with original dataframe

```python
movies_full = movies3.merge(genre_counts_df,
                           on='Major_Genre',
                           # right_on='Genre',
                           how='left')

movies_full.head(10)
```

### Adding info to map

```python
state_map = alt.topo_feature(data.us_10m.url, 'states')
alt.Chart(state_map).mark_geoshape().encode(color='id:Q').project(type='albersUsa')
```

```python
income =  data.income()
income
```

```python
alt.Chart(state_map).mark_geoshape(
).transform_lookup(
    lookup='id',
    from_=alt.LookupData(income, 'id', ['pct', 'name', 'region'])
).encode(color='pct:Q',
        tooltip=['name:N', 'region:N']).project(type='albersUsa')
```
